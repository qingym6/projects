---

### **Task 2: Statistical Hypothesis Testing**

To evaluate the relationships observed during EDA, three statistical tests were applied to address the **challenge goal of Statistical Hypothesis Testing**.

1. **ANOVA (BMI vs Diet Category)**  
   - **Purpose:** Determine whether mean BMI differs among individuals with different diet recommendations.  
   - **Verification:** Checked normality within groups (Shapiro–Wilk) and equality of variances (Levene’s test).  
   - **Outputs:** ANOVA table with F-statistic and p-value to evaluate group differences.

2. ***t*-Test (Age: Low Sodium vs Others)**  
   - **Purpose:** Compare mean ages between individuals recommended a Low_Sodium diet and those with other diets.  
   - **Verification:** Tested normality for both groups and equal variances; used Welch’s *t*-test if variances differed.  
   - **Outputs:** t-statistic, p-value, and interpretation of group mean differences.

3. **Chi-Square Test (Physical Activity vs Diet)**  
   - **Purpose:** Examine whether physical activity level is associated with diet category.  
   - **Verification:** Ensured all expected frequencies ≥ 5 and visualized observed proportions.  
   - **Outputs:** Chi-square statistic, degrees of freedom, and p-value, with Cramér’s V for association strength.

```python
def missing_summary(df, name):
    """
    Takes a DataFrame and its name, prints a summary dictionary of missing (NaN) values
    including total count, percentage, and columns with missing data; returns None.

    >>> test_df = pd.DataFrame({
    ...     'age': [20, np.nan, 30],
    ...     'bmi': [22.1, 23.4, 25.0],
    ...     'disease': [None, 'Diabetes', None]
    ... })
    >>> missing_summary(test_df, "Test Dataset")
    Missing Data Summary for Test Dataset
    Total missing values: 3
    Percent missing (overall): 33.33%
    Columns with missing values:
    {'age': 1, 'disease': 2}
    """
    print(f"Missing Data Summary for {name}")
    
    # Count missing values per column
    missing = df.isna().sum()
    
    # Keep only columns where missing count > 0
    missing = missing[missing > 0]
    
    if missing.empty:
        print("No missing data found!\n")
    else:
        total_cells = df.shape[0] * df.shape[1]
        total_missing = missing.sum()
        percent_missing = (total_missing / total_cells) * 100

        print(f"Total missing values: {int(total_missing)}")
        print(f"Percent missing (overall): {percent_missing:.2f}%")
        print("Columns with missing values:")
        print(missing.to_dict())
    
    return None

doctest.run_docstring_examples(missing_summary, globals())

# Apply to both datasets
missing_summary(diet_rec, "Diet Recommendation Dataset")
missing_summary(nutrient, "Nutrient Composition Dataset")
```

### Result Validity only
**State your null hypothesis for each research question you intend to do result validation on. Give the type of statistical test you will be using and the assumptions for that test. As part of your EDA, you must verify the assumptions for your test.**

**Research Question:**  
Is there a statistically significant difference in BMI among individuals recommended Balanced, Low_Carb, and Low_Sodium diets?

**Null Hypothesis (H₀):**  
μ_Balanced = μ_Low_Carb = μ_Low_Sodium  

**Alternative Hypothesis (H₁):**  
At least one group's mean BMI differs from the others.  

**Planned Test:**  
One-way ANOVA with `Diet_Recommendation` as the factor and `BMI` as the response.

**Assumptions:**
- **Independence:** Each individual’s BMI is independent of others.  
- **Normality:** BMI values are approximately normally distributed within each diet category.  
- **Equal Variance:** The variance of BMI is similar across the three diet groups.

**Verification Steps:**
- Visualize BMI distributions with histograms and boxplots to assess symmetry and outliers. (Already shown in the Visualization part)
- Compute group-wise statistics; run **Shapiro–Wilk** (normality) and **Levene’s test** (equal variance).  
- Confirm reasonable sample sizes so ANOVA remains robust to mild violations.


```python
def test_normality_by_group(df, group_col, value_col):
    """
    Runs the Shapiro–Wilk normality test for a given numeric variable
    across groups in a categorical column.

    Parameters
    ----------
    df : pandas.DataFrame
        The dataset containing both the grouping and numeric columns.
    group_col : str
        The name of the column representing groups (categorical variable).
    value_col : str
        The name of the numeric column to test for normality.

    Returns
    -------
    dict
        Dictionary of group names mapped to (W-statistic, p-value) tuples.

    Example
    -------
    >>> data = pd.DataFrame({
    ...     'Group': ['A']*5 + ['B']*5,
    ...     'Value': [22, 23, 21, 22, 24, 29, 30, 28, 27, 31]
    ... })
    >>> results = test_normality_by_group(data, 'Group', 'Value')
    Shapiro–Wilk Normality Test Results for 'Value' grouped by 'Group':
    <BLANKLINE>
    A: W=0.961, p=0.8140
    B: W=0.987, p=0.9672
    >>> isinstance(results, dict)
    True
    """
    print(f"Shapiro–Wilk Normality Test Results for '{value_col}' grouped by '{group_col}':\n")
    results = {}
    for group, subset in df.groupby(group_col):
        values = subset[value_col].dropna()
        if len(values) < 3:
            print(f"{group}: Not enough data (n={len(values)}) to run Shapiro–Wilk test.")
            continue
        stat, p = shapiro(values)
        results[group] = (stat, p)
        print(f"{group}: W={stat:.3f}, p={p:.4f}")
    return results


def test_equal_variance(df, group_col, value_col):
    """
    Runs Levene’s test to assess whether the variances of a numeric variable
    are equal across different groups.

    Parameters
    ----------
    df : pandas.DataFrame
        The dataset containing both the grouping and numeric columns.
    group_col : str
        The name of the column representing groups (categorical variable).
    value_col : str
        The name of the numeric column to test for variance equality.

    Returns
    -------
    tuple
        (W-statistic, p-value)

    Example
    -------
    >>> data = pd.DataFrame({
    ...     'Group': ['A']*5 + ['B']*5 + ['C']*5,
    ...     'Value': [22, 23, 21, 22, 24, 29, 30, 28, 27, 31, 26, 25, 27, 26, 25]
    ... })
    >>> stat, p = test_equal_variance(data, 'Group', 'Value')
    Levene’s Test for Equal Variances on 'Value' by 'Group':
    W=0.824, p=0.4622
    >>> isinstance(stat, float) and isinstance(p, float)
    True
    """
    groups = [subset[value_col].dropna() for _, subset in df.groupby(group_col)]
    stat, p = levene(*groups)
    print(f"Levene’s Test for Equal Variances on '{value_col}' by '{group_col}':")
    print(f"W={stat:.3f}, p={p:.4f}")
    return stat, p

doctest.run_docstring_examples(test_normality_by_group, globals())
doctest.run_docstring_examples(test_equal_variance, globals())

# Test on real dataset (BMI by Diet Recommendation)
test_normality_by_group(diet_rec, "Diet_Recommendation", "BMI")
print("\n")
test_equal_variance(diet_rec, "Diet_Recommendation", "BMI")
```

**Verification Conclusion:**  
Shapiro–Wilk tests indicated slight non-normality in BMI distributions across diet categories (p < 0.05). However, given adequate group sizes (Each group has a reasonably large sample size (n ≥ 30), and sample sizes across groups are roughly equal) and the robustness of ANOVA to minor deviations, the test remains appropriate. Levene’s test confirmed equal variances (p = 0.4154), satisfying the homogeneity assumption.

**Research Question:**  
Is physical activity level associated with diet recommendation categories?

**Null Hypothesis (H₀):**  
Physical activity level is **independent** of diet recommendation category.  

**Alternative Hypothesis (H₁):**  
Physical activity level and diet recommendation category are **associated** (not independent).  

**Planned Test:**  
Chi-square test of independence on a contingency table between `Physical_Activity_Level` and `Diet_Recommendation`.

**Assumptions:**
- **Independence of Observations:** Each participant contributes to exactly one cell in the contingency table.  
- **Expected Cell Counts:** Most expected frequencies are ≥ 5; if not, categories may need to be combined to meet this condition.  

**Verification Steps:**
- Construct and inspect the contingency table (`pd.crosstab`).  
- Check that all expected counts are ≥ 5 to validate chi-square assumptions.  
- Visualize the relationship using a grouped bar plot to identify differences in activity level proportions across diet types.  
- If assumptions are violated, consider using **Fisher’s Exact Test** or combining sparse categories.  


```python
def verify_chisq_assumptions(df):
    """
    Verifies Chi-square test assumptions between Physical_Activity_Level and Diet_Recommendation.
    """
    # Contingency Table
    contingency = pd.crosstab(df["Physical_Activity_Level"], df["Diet_Recommendation"])
    print("Contingency Table (Observed Counts):")
    display(contingency)

    # Expected Counts
    chi2, p, dof, expected = chi2_contingency(contingency)
    expected_df = pd.DataFrame(expected, 
                               index=contingency.index, 
                               columns=contingency.columns)
    print("\nExpected Frequencies:")
    display(expected_df.round(2))

    # Check for Small Expected Counts
    min_expected = expected_df.values.min()
    print(f"\nMinimum Expected Count: {min_expected:.2f}")
    if min_expected < 5:
        print("Warning: Some cells have expected counts < 5. Consider combining categories.")
    else:
        print("All expected counts ≥ 5. Chi-square assumptions are satisfied.")

    # Visualization
    contingency_norm = contingency.div(contingency.sum(axis=1), axis=0)
    contingency_norm.plot(kind="bar", figsize=(8, 5), colormap="Set2")
    plt.title("Physical Activity Level vs Diet Recommendation")
    plt.xlabel("Physical Activity Level")
    plt.ylabel("Proportion of Individuals")
    plt.legend(title="Diet Category")
    plt.tight_layout()
    plt.show()
```

```python
# Test Section for verify_chisq_assumptions()

# Create a small synthetic dataset for testing
test_df = pd.DataFrame({
    "Physical_Activity_Level": [
        "Low", "Low", "Medium", "Medium", "High", "High",
        "Low", "Medium", "High", "Low", "High", "Medium"
    ],
    "Diet_Recommendation": [
        "Balanced", "Low_Carb", "Low_Sodium", "Balanced", "Low_Carb", "Balanced",
        "Low_Carb", "Low_Sodium", "Balanced", "Low_Sodium", "Low_Carb", "Balanced"
    ]
})

# Run the verification function
verify_chisq_assumptions(test_df)

# Assertions to validate behavior
# Check if a contingency table is generated correctly
contingency = pd.crosstab(test_df["Physical_Activity_Level"], test_df["Diet_Recommendation"])
assert isinstance(contingency, pd.DataFrame), "Contingency table should be a DataFrame."
assert contingency.shape[0] > 0 and contingency.shape[1] > 0, "Contingency table must have data."

# Check Chi-square expected frequencies
chi2, p, dof, expected = chi2_contingency(contingency)
expected_df = pd.DataFrame(expected, index=contingency.index, columns=contingency.columns)
assert expected_df.shape == contingency.shape, "Expected frequency table shape should match contingency table."

# Check that all expected counts are positive
assert (expected_df.values >= 0).all(), "Expected frequencies must be non-negative."

# Ensure plot renders properly (Axes object created)
ax = contingency.div(contingency.sum(axis=1), axis=0).plot(kind="bar", figsize=(6, 4))
assert ax.get_title() == "", "The test plot should render correctly without title mismatch."
plt.close(ax.figure)  # Close plot to avoid multiple pop-ups during testing

print("\n All tests passed for verify_chisq_assumptions()!")
```

**Verification Conclusion:**

All cells in the expected frequency table have values greater than 5 (minimum expected count = 85.14), which satisfies the Chi-square assumption for sufficient expected counts.

Therefore, both major assumptions for the Chi-square test — independence of observations and adequate expected cell frequencies — are satisfied. This means the Chi-square test of independence can be appropriately applied to examine whether `Physical_Activity_Level` and `Diet_Recommendation` are statistically associated.
