# Recommending Personalized Diet Categories and Foods Based on Nutrient Needs

This project aims to explore how data-driven methods can provide personalized diet recommendations that match individuals’ demographic and lifestyle characteristics with foods that help them meet nutrient targets.

### Research Questions
1. **Are there statistically significant differences in demographic and lifestyle features among different diet recommendation groups?**  
   → I will use ANOVA, *t*-tests, and chi-square tests to test whether variables such as BMI, age, and activity level vary significantly across diet groups.

2. **Can we predict the best diet recommendation for an individual based on demographic and lifestyle features (e.g., age, physical activity, BMI, diet type)?**  
   → I will train machine-learning classifiers (kNN, Decision Tree, and SVM) and compare their performance on prediction metrics.

3. **What are the top foods that can be recommended to help individuals meet their nutrient targets in each category of nutrients?**  
   → I will use the nutrient-composition dataset to rank foods by how well they fit each diet category’s nutrient requirements.

4. **Which machine-learning algorithm performs best for personalized diet recommendations?**  
   → I will tune and evaluate each model using accuracy, precision, recall, F1-score, and ROC-AUC to identify the most effective one.

```python
def missing_summary(df, name):
    """
    Takes a DataFrame and its name, prints a summary dictionary of missing (NaN) values
    including total count, percentage, and columns with missing data; returns None.

    >>> test_df = pd.DataFrame({
    ...     'age': [20, np.nan, 30],
    ...     'bmi': [22.1, 23.4, 25.0],
    ...     'disease': [None, 'Diabetes', None]
    ... })
    >>> missing_summary(test_df, "Test Dataset")
    Missing Data Summary for Test Dataset
    Total missing values: 3
    Percent missing (overall): 33.33%
    Columns with missing values:
    {'age': 1, 'disease': 2}
    """
    print(f"Missing Data Summary for {name}")
    
    # Count missing values per column
    missing = df.isna().sum()
    
    # Keep only columns where missing count > 0
    missing = missing[missing > 0]
    
    if missing.empty:
        print("No missing data found!\n")
    else:
        total_cells = df.shape[0] * df.shape[1]
        total_missing = missing.sum()
        percent_missing = (total_missing / total_cells) * 100

        print(f"Total missing values: {int(total_missing)}")
        print(f"Percent missing (overall): {percent_missing:.2f}%")
        print("Columns with missing values:")
        print(missing.to_dict())
    
    return None

doctest.run_docstring_examples(missing_summary, globals())

# Apply to both datasets
missing_summary(diet_rec, "Diet Recommendation Dataset")
missing_summary(nutrient, "Nutrient Composition Dataset")
```

### Statistical Hypothesis Testing

#### **1. ANOVA**
**Question:** Is there a statistically significant difference in BMI among individuals recommended *Balanced*, *Low_Carb*, and *Low_Sodium* diets?  

**Null hypothesis (H₀):**  
μ<sub>Balanced</sub> = μ<sub>Low Carb</sub> = μ<sub>Low Sodium</sub>  

**Alternative hypothesis (H₁):**  
At least one group mean differs.

---

#### **2. *t*-Test**
**Question:** Are older individuals more likely to be recommended low-sodium diets?  

**Null hypothesis (H₀):**  
μ<sub>Low Sodium</sub> = μ<sub>Others</sub>  

**Alternative hypothesis (H₁):**  
μ<sub>Low Sodium</sub> ≠ μ<sub>Others</sub>

---

#### **3. Chi-Square Test**
**Question:** Is the physical activity level associated with the diet recommendation categories?  

**Null hypothesis (H₀):**  
Activity level is independent of diet recommendation category.  

**Alternative hypothesis (H₁):**  
There is an association between activity level and diet recommendation category.
